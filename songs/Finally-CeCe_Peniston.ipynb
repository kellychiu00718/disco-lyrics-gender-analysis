```python
import os
import re
import nltk
import nltk.corpus
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.tokenize import RegexpTokenizer
import pandas as pd
import plotly.express as px

file = open('Finally-CeCe Peniston.txt', 'r')
content= file.read()
print(content)
```

```python
content_tokens=word_tokenize(content)
print(content_tokens)
print(len(content_tokens))
```

```python
#freq of original tokenized lyrics

freq=FreqDist()

for word in content_tokens:
    freq[word.lower()]+=1
    
freq
```

```python
from nltk.tokenize import blankline_tokenize
content_blankline=blankline_tokenize(content)
print(len(content_blankline))

# print(content_blankline[0])
```

```python
punctuation= re.compile(r"[-.,''``?!:;()|0-9]")

post_punctuation=[]
for x in content_tokens:
    word=punctuation.sub("", x)
    if len(word)>0:
        post_punctuation.append(word)
        
post_punctuation #new tokens after removing -.,'?!:;()|0-9
```

```python
new_freq=FreqDist()

for word in post_punctuation:
    new_freq[word.lower()]+=1
    
print(new_freq) #words frequency without "-.,'?!:;()|0-9"

print(len(new_freq))

new_freq_top10=new_freq.most_common(10)
print(new_freq_top10)
```

```python
from nltk.corpus import stopwords
stop_words=set(stopwords.words('english'))

filtered_sentence=[] #sentence without stopwords(meaningless words)

for word in post_punctuation:
    if word not in stop_words:
        filtered_sentence.append(word)
        
filtered_sentence #sentence without stopwords(meaningless words)
```

```python
freq_without_stopwords=FreqDist()

for word in filtered_sentence:
    freq_without_stopwords[word.lower()]+=1
    
print(freq_without_stopwords) #words frequency without stopwords

print(len(freq_without_stopwords))

freqStopwords_top10=freq_without_stopwords.most_common(10)
print(freqStopwords_top10)
```

```python
pos_tagged=[]
for word in post_punctuation:
    pos_tagged.append(nltk.pos_tag([word]))
    
pos_tagged #part-of-speech tokens without stopwords
```

```python
from nltk import ne_chunk

tags=[]
for x in pos_tagged:
    for y in x:
        tags.append(y)

ne_chunked=ne_chunk(tags)

# extract all named entities
named_entities=[]
for x in ne_chunked:
    if hasattr(x, 'label'):
        entity_name=' '.join(c[0] for c in x.leaves())
        entity_type=x.label()
        named_entities.append((entity_name, entity_type))

named_entities #nothing in named entity recognition
ne_chunked
```

```python
grammar_np= r"NP: {<DT>?<JJ>*<NN>}" #NOUN PARSER

chunk_parser= nltk.RegexpParser(grammar_np)

chunk_result=chunk_parser.parse(tags)

chunk_result
```

```python
grammar_np= r"VP: {<PRP>?<VB|VBD|VBZ|VBG>*<RB|RBR>?}" #VERB PARSER

chunk_parser= nltk.RegexpParser(grammar_np)

chunk_result=chunk_parser.parse(tags)

chunk_result
```

```python
import plotly.express as px

freq_list_=list(freq_without_stopwords.items())
dic={"word":[], "freq":[]}
for x, y in freq_list_:
    dic["word"].append(x)
    dic["freq"].append(y)


# plotting the bar chart
fig = px.bar(dic, x='freq', y='word', title= "The frequency of words in 'Finally(without stopwords)'", height=1500, orientation='h')
fig.update_layout(bargap=0.7, yaxis={'categoryorder':'total ascending'})
fig.update_traces(width=1)

# showing the plot
fig.show()
```
