```python
import os
import re
import nltk
import nltk.corpus
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.tokenize import RegexpTokenizer
import pandas as pd
import plotly.express as px

file = open('i will survive_lyrics.txt', 'r')
content= file.read()
print(content)
```

```python
from nltk.corpus import stopwords
len(stopwords.words('english'))
```

```python
punctuation= re.compile(r"[-.,'?!:;()|0-9]")

post_punctuation=[]
for x in content_tokens:
    word=punctuation.sub("", x)
    if len(word)>0:
        post_punctuation.append(word)
        
post_punctuation #new tokens after removing -.,'?!:;()|0-9
```

```python
new_freq=FreqDist()

for word in post_punctuation:
    new_freq[word.lower()]+=1
    
new_freq #words frequency without "-.,'?!:;()|0-9"
```

```python
len(new_freq)
```

```python
new_freq_top10=new_freq.most_common(10)
new_freq_top10 
```

```python
stop_words=set(stopwords.words('english'))

filtered_sentence=[] #sentence without stopwords(meaningless words)

for word in post_punctuation:
    if word not in stop_words:
        filtered_sentence.append(word)
        
filtered_sentence #sentence without stopwords(meaningless words)
```

```python
freq_without_stopwords=FreqDist()

for word in filtered_sentence:
    freq_without_stopwords[word.lower()]+=1
    
freq_without_stopwords #words frequency without stopwords
```

```python
len(freq_without_stopwords)
```

```python
freqStopwords_top10=freq_without_stopwords.most_common(10)
freqStopwords_top10
```

Step-by-step:

1. tokenize
2. remove Punctuation 
3. remove stopwords
->calculate frequency, the top words used, the topics lying under the lyrics.... 

with original lyrics: whole lyrics-> paragraph by paragraph-> lines by lines

4. part-of-speech tag
5. named entity recognition tag
6. chunking
7. sentiment analysis 

```python
#the first paragraph analysis
paragraph_one='''At first I was afraid, I was petrified.
Kept thinkin' I could never live without you by my side.
But then I spent so many nights thinkin' how you did me wrong.
And I grew strong and I learned how to get along.
'''
paragraph_one_tokens=word_tokenize(paragraph_one) #tokenize
```

```python
punctuation= re.compile(r"[-.,'?!:;()|0-9]")

post_paragraph_one=[]
for x in paragraph_one_tokens:
    word=punctuation.sub("", x)
    if len(word)>0:
        post_paragraph_one.append(word)
        
post_paragraph_one #new tokens after removing -.,'?!:;()|0-9
```

```python
stop_words=set(stopwords.words('english'))

filtered_paragraph_one=[] #sentence without stopwords(meaningless words)

for word in post_paragraph_one:
    if word not in stop_words:
        filtered_paragraph_one.append(word)
        
filtered_paragraph_one #sentence without stopwords(meaningless words)
```

```python
pos_paragraph_one_tagged=[]
for word in post_paragraph_one:
    pos_paragraph_one_tagged.append(nltk.pos_tag([word]))
    
pos_paragraph_one_tagged #part-of-speech tokens without punctuation
```

```python
from nltk import ne_chunk

paragraph_one_tags=[]
for x in pos_paragraph_one_tagged:
    for y in x:
        paragraph_one_tags.append(y)

ne_chunked=ne_chunk(paragraph_one_tags)

# extract all named entities
named_entities=[]
for x in ne_chunked:
    if hasattr(x, 'label'):
        entity_name=' '.join(c[0] for c in x.leaves())
        entity_type=x.label()
        named_entities.append((entity_name, entity_type))

named_entities #nothing in named entity recognition
ne_chunked
```

```python
grammar_np= r"NP: {<DT>?<JJ>*<NN>}" #NOUN PARSER

chunk_parser= nltk.RegexpParser(grammar_np)

chunk_result=chunk_parser.parse(paragraph_one_tags)

chunk_result
```

```python
grammar_np= r"VP: {<PRP>?<VB|VBD|VBZ|VBG>*<RB|RBR>?}" #VERB PARSER

chunk_parser= nltk.RegexpParser(grammar_np)

chunk_result=chunk_parser.parse(paragraph_one_tags)

chunk_result
```

```python
filtered_paragraph_one

freq_paragraph_one=FreqDist()

for word in filtered_paragraph_one:
    freq_paragraph_one[word.lower()]+=1
    
freq_paragraph_one 
```

```python
import plotly.express as px

freq_list_=list(freq_paragraph_one .items())
print(len(freq_list_))
dic={"word":[], "freq":[]}
for x, y in freq_list_:
    dic["word"].append(x)
    dic["freq"].append(y)


# plotting the bar chart
fig = px.bar(dic, x='word', y='freq', title= "The frequency of words in 'I Will Survive(without stopwords)'", height=400)
fig.update_layout(bargap=0.7, xaxis={'categoryorder':'total descending'})
fig.update_traces(width=1)

# showing the plot
fig.show()
```
